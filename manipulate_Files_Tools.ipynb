```python
import sqlite3
import json
print("test")
# 1. Connexion à la base de données extraite
# Note : Il faut d'abord renommer le fichier extrait 'collection.anki2' en 'collection.db'
conn = sqlite3.connect('collection.db')
cursor = conn.cursor()

# 2. Récupérer les notes (les données brutes des kanjis)
# Dans Anki, le texte est stocké dans le champ 'flds' de la table 'notes'
cursor.execute("SELECT flds FROM notes")
notes = cursor.fetchall()

kanji_data = []

for note in notes:
    # Les champs sont séparés par le caractère '\x1f'
    fields = note[0].split('\x1f')
    kanji_data.append({
        "departement": fields[0],
        "keyword": fields[1],
        "image":fields[2],
        "mot-cle":fields[3],
        "histoire":fields[4],
        "remarques":fields[5] if len(fields) > 5 else ""
    })

# 3. Sauvegarder en JSON pour votre chatbot
with open('departement.json', 'w', encoding='utf-8') as f:
    json.dump(kanji_data, f, ensure_ascii=False, indent=4)

print(f"{len(kanji_data)} departements extraits avec succès !")
```

```python
import os
import csv

# --- 設定項目 ---
# 1. リネームしたいファイルが入っているフォルダのパス
target_folder = r'C:\Users\owner\Documents\French_Departments_Map_Insee_Flag_Capital'

# 2. リネーム対応表（CSV）のパス
csv_path = r'C:\Users\owner\Documents\kanjilock-web\tools\correction01.csv'

def rename_files(preview=True):
    # CSVファイルを読み込む
    if not os.path.exists(csv_path):
        print(f"エラー: CSVファイルが見つかりません: {csv_path}")
        return

    with open(csv_path, mode='r', encoding='utf-8') as f:
        # CSVの形式に合わせて調整（ヘッダーがない場合はこのまま）
        reader = csv.reader(f)
        
        rename_count = 0
        skip_count = 0

        print(f"{'--- プレビューモード ---' if preview else '--- 実行モード ---'}")
        
        for row in reader:
            if len(row) < 2:
                continue
            
            old_name = row[0].strip()
            new_name = row[1].strip()
            
            old_path = os.path.join(target_folder, old_name)
            new_path = os.path.join(target_folder, new_name)
            
            # 元のファイルが存在するか確認
            if os.path.exists(old_path):
                if preview:
                    print(f"[PREVIEW] {old_name} -> {new_name}")
                else:
                    try:
                        os.rename(old_path, new_path)
                        print(f"[SUCCESS] {old_name} -> {new_name}")
                    except Exception as e:
                        print(f"[ERROR] {old_name} の変更に失敗しました: {e}")
                rename_count += 1
            else:
                print(f"[SKIP] 元ファイルが見つかりません: {old_name}")
                skip_count += 1

        print(f"\n完了: {rename_count}件を対象、{skip_count}件をスキップしました。")
        if preview:
            print("注意: まだ名前は変更されていません。確認して問題なければ、rename_files(preview=False) を実行してください。")

# --- 実行 ---
# まずは確認（プレビュー）
rename_files(preview=False)
```
