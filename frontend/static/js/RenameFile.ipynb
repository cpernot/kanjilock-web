```python
import os
import csv

# --- 設定 ---
folder_path = r"C:\Users\owner\Documents\French_Departments_Map_Insee_Flag_Capital"
csv_file = r"C:\Users\owner\Documents\rename_departments.csv"  # CSVファイルのパス

def rename_files():
    if not os.path.exists(folder_path):
        print(f"❌ フォルダが見つかりません: {folder_path}")
        return

    try:
        with open(csv_file, mode='r', encoding='utf-8-sig') as f:
            reader = csv.reader(f)
            
            count = 0
            for row in reader:
                # 1列目 (index 0) が現在の名前、3列目 (index 2) が新しい名前
                if len(row) < 3:
                    continue
                
                old_name = row[0].strip()
                new_name = row[2].strip()
                
                old_full_path = os.path.join(folder_path, old_name)
                new_full_path = os.path.join(folder_path, new_name)
                
                # ファイルが存在するか確認してリネーム
                if os.path.exists(old_full_path):
                    try:
                        os.rename(old_full_path, new_full_path)
                        print(f"✅ 変更済: {old_name} -> {new_name}")
                        count += 1
                    except Exception as e:
                        print(f"⚠️ エラー ({old_name}): {e}")
                else:
                    print(f"❓ 見つかりません: {old_name}")
            
            print(f"\n✨ 完了！合計 {count} 個のファイルをリネームしました。")

    except Exception as e:
        print(f"❌ CSV読み込みエラー: {e}")

rename_files()
```
